```python
#Ques 27 - Write a function to check if a number is Armstrong or not.
num = int(input("Enter the number: "))
copy_num = num
order = len(str(num))
def arms_numb(num):
    sum = 0
    while num > 0:
       digit = num % 10
       sum = sum + digit ** order
       num = num // 10
    if sum == copy_num:
       print("Armstrong Number")
    else:
       print("Not Armstrong Number")
arms_numb(num)
```

```python
 #Ques - 5. Write a function that returns the GCD of two numbers.
n1 = int(input("Enter first number: "))
n2 = int(input("Enter second number: "))

def Hcf_find(n1,n2):
    if n2>n1:
       mn = n1
    else:
       mn = n2

    for i in range(1,mn+1):
        if n1%i==0 and n2%i==0:
           hcf = i
           lcm = (n1 * n2)/hcf
    print(f"HCF/GDC of {hcf}")
    print(f"LCM of {lcm}")
Hcf_find(n1,n2)
```
